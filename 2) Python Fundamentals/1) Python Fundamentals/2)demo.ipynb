```python
l = [1, 'Hello', None, 2, 4]

res = 0

for i in l:
    if type(i) == int:
        res += i

print('The output is :',res)
```

# String Manipulation Functions

```python
name='mangesh sanjay rajguru'
city='PUNE'
designation='   Software Developer   '
data = "Python,SQL,Spark"
```

```python
print(len(name))    # 1) len() 
print(name.upper()) # 2) Upper() 
print(city.lower()) # 3) lower()
print(city.capitalize())   # 4) Capitalize()
print(name.title())        # 5) Title()
print(designation.strip()) # 6)strip ()
name=name.replace('mangesh','durga')
print(name)  # 7) replace()

# 8) Split() function--> Converts the string into a list
data.split(",") # data.split(",")[1] returns element at position 1 in the list

# 9) Join() function--> Exactly opposite to the Split() function
result=",".join(data)
print(result)

# 10) Find()  -->returns the index of the string
print(name.find("sanjay"))

# 11) Count() -->returns the count of string, that how many times is it repeated
print(name.count('a'))
```
