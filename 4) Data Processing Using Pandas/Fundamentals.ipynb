#### 2) Create Dataframe 

```python
# creating sample dataframes
data={
    "name": ["Mangesh", "Rahul", "Amit","Pratik","Aryan","Eshaan"],
    "age": [20, 21, 22,20,19,23],
    "city": ["Pune", "Mumbai", "Nashik","Gujrat","Kashmir","Jammu"],
    "marks":[96,78,71,93,86,81]
}

df=pd.DataFrame(data)
print(df)


# loading dataframes into csv, json and excel files
df.to_csv("output.csv",index=False)
df.to_excel("output2.xlsx",index=False)
df.to_json("output3.json",index=False)
```

#### 4) Rows & Columns - Selection

```python
df[['name']] ## select single column
```

```python
df[['name','age']] #selects multiple column
```

```python
# returns multiple rows
df.loc[0:2]
```

```python
# selects single row || loc - label based index
df.loc[df.name=="Rahul"]
```

```python
# select multiple rows with conditions || loc - label based index
df.loc[(df.name=="Mangesh") & (df.age>=20)]
```

#### 5) Filter Dataframe - Filtering by column values

```python
# filter people having age more than equal to 20 into a new dataframe
df_age_filter=df[df['age']>=21]
df_age_filter
```

```python
# multiple filter condition
df_multiple_filters=df[(df['age']>=22) & (df['marks']>=80)]
df_multiple_filters
```

```python
# where function replace values in a DataFrame based on a condition

df.where(df['age']>21,other='NAN')
```
