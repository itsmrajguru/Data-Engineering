# Overview of JSON Strings and Files

## 1. What is JSON?

* **JSON = JavaScript Object Notation**
* JSON is a **data format** used to represent and exchange structured data.
* Commonly used with **APIs, files, and data pipelines**.
* JSON mainly represents:

  * **Objects** → similar to Python `dict`
  * **Arrays** → similar to Python `list`

---

## 2. JSON Object

A JSON object contains **key-value pairs**.

Example:

```json
{
    "order_id": 1,
    "order_date": "2013-07-25"
}
```

It looks similar to a Python dictionary:

```python
{
    "order_id": 1,
    "order_date": "2013-07-25"
}
```

But JSON and Python `dict` are **not the same thing**.

---

## 3. JSON Object vs JSON Array

| JSON              | Similar Python Type |
| ----------------- | ------------------- |
| JSON Object `{ }` | `dict`              |
| JSON Array `[ ]`  | `list`              |

Example:

```text
JSON Object
     ↓
{ "name": "Mangesh" }
     ↓
Python dict
```

```text
JSON Array
     ↓
["Python", "SQL", "Spark"]
     ↓
Python list
```

**But raw JSON can initially be stored as a Python `str`.**

---


## 4. Why Do We Need to Convert JSON?

Suppose we receive JSON data from an API:

```text
API
 ↓
JSON data
 ↓
Python string
```

We eventually do:

```text
Python string
      ↓
   Parse JSON
      ↓
Python dict/list
      ↓
Process the data
```

Once JSON is converted into a Python `dict` or `list`, we can easily access and process its values.

---

## 5. Serialization and Deserialization

### Deserialization

Converting JSON data into Python objects:

```text
JSON
  ↓
Python dict / list
```

### Serialization

Converting Python objects into JSON:

```text
Python dict / list
  ↓
JSON
```

### Easy Way to Remember

```text
Deserialize → JSON → Python

Serialize → Python → JSON
```
