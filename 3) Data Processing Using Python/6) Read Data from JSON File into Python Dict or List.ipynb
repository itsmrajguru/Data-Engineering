# Now Lets see , How to actually read data from JSON file into Python Dict or list

````python
# Step 1 :Import Json
import json
````

````python
# Step 2 :json.loads() — JSON string(Object) → Python Object(dict)
order_jd = '{"order_id": 1, "order_date": "2013-07-25"}'
order=json.loads(order_jd)
print(order["order_date"])
````

````python
# Step 2 :json.loads() — JSON string(Array) → Python Object(list)
cities_ja = '["Dallas", "Frisco", "Hyderabad"]'
cities=json.loads(cities_ja)
print(cities[1])
````

### Note :Why do we need Loads ?<b> Because loading the JSON string using loads, it was a string and we have to convert it into a python list or dict , so that we load it 

### 1) json.loads() vs json.load() 

- json.loads() → JSON STRING → Python object

- json.load()  → JSON FILE   → Python object

### 2) json.loads() — Read JSON String

- Here the JSON String is converted into Python Dict
```python

        import json

        data = '{"name": "Mangesh", "age": 20}'

        result = json.loads(data)

        print(result)

```

### 3) json.load() — Read JSON File

````python
import json
with open("data/retail_db/schemas.json") as f:
    schemas = json.load(f)

print(schemas.keys())
print(schemas.values())
````
