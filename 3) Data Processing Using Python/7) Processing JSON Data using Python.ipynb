# Now lets see , How to process the JSON Files using python


- How to get the nested columns details like <br>
 ```['order_id', 'order_date', 'order_customer_id', 'order_status']```<br>as a output from the below database ?

```python
    column_details = [
        {'column_name': 'order_id', 'data_type': 'integer', 'column_position': 1},
        {'column_name': 'order_date', 'data_type': 'string', 'column_position': 2},
        {'column_name': 'order_customer_id','data_type': 'integer','column_position': 3},
        {'column_name': 'order_status', 'data_type': 'string', 'column_position': 4}
    ]
```


````python
# lets import json first
import json

# import the file using file_path
file_path = 'data/retail_db/schemas.json'

# load the data (JSON file)
schemas=json.load(open(file_path))

# get column details for orders table
column_details=schemas["orders"]

# Sort columns according to their position
columns = sorted(
    column_details,
    key=lambda col: col['column_position']
)

# Extract only column names
column_names = [col['column_name'] for col in columns]

print(column_names)
````

````python
# This whole task can be done by using one single function
list(map(lambda col: col['column_name'], column_details))
````
