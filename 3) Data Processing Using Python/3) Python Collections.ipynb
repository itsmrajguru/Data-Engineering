# Python Collections

Python provides four commonly used collection types:

* `list`
* `set`
* `tuple`
* `dict`

---

## 1. list

* **Ordered collection**
* **Duplicates allowed**
* **Mutable** → values can be changed
* Supports **indexing**
* Created using `[]`

### Example

```python
students = ["Mangesh", "Rahul", "Amit", "Rahul"]

print(students[0])       # Mangesh
students[0] = "Rohit"    # Value can be changed
```

---

## 2. set

* Collection of **unique values**
* **Duplicates are automatically removed**
* Mutable
* Does **not support indexing**
* Created using `{}`

### Example

```python
cities = {"Pune", "Mumbai", "Pune", "Delhi"}

print(cities)
# {'Pune', 'Mumbai', 'Delhi'}
```

**Main use:** Remove duplicates / get unique values.

---

## 3. tuple

* **Ordered collection**
* Duplicates allowed
* **Immutable** → values cannot be changed
* Supports indexing
* Created using `()`

### Example

```python
student = ("Mangesh", 20, "Pune")

print(student[0])    # Mangesh

# student[0] = "Rahul"   # Error - tuple cannot be changed
```

**Main use:** Store values that should remain unchanged.

---

## 4. dict

* Collection of **key-value pairs**
* Each **key must be unique**
* Mutable
* Values are accessed using their **keys**
* Created using `{key: value}`

### Example

```python
student = {
    "name": "Mangesh",
    "age": 20,
    "city": "Pune"
}

print(student["name"])   # Mangesh
print(student["age"])    # 20
```

---

# Quick Difference

| Collection | Ordered | Duplicates | Mutable | Access      |
| ---------- | ------- | ---------- | ------- | ----------- |
| `list`     | Yes     | Yes        | Yes     | Index       |
| `set`      | No*     | No         | Yes     | No indexing |
| `tuple`    | Yes     | Yes        | No      | Index       |
| `dict`     | Yes**   | Keys: No   | Yes     | Key         |

* Don't rely on set ordering.
** Dictionaries preserve insertion order in modern Python.

---

# Easy Way to Remember

```text
list  → Ordered + Duplicates + Changeable
set   → Unique values
tuple → Ordered + Cannot change
dict  → Key → Value
```

### Examples

```python
# list
["Python", "SQL", "Python"]

# set
{"Python", "SQL"}

# tuple
("Mangesh", 20, "Pune")

# dict
{"name": "Mangesh", "age": 20}
```
