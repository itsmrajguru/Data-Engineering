### 1) Lambda Functions vs  Normal Funtions

````python
# 1) Normal functions 
def sum(n):
    return (n*(n+1))/2

#lets create a python list
l=[1,2,3,4] # and you want output as l=[1,3,6,10]
result=[]

# Way 1 :Normal Way
for i in l:
    result.append(sum(i))
print(result)

# Way 2: List comprehension

[sum(n) for n in l]
````

````python
# 2) lambda Functions
""" 1-> No name
    2-> No return type
    3-> ONly one statent can be executed as a return type after :
    4-> """ 
    
sum=lambda n:(n*(n+1))/2
sum(5)
````

### 2) Map( ) Function

- Simply , Map function is used to apply a function to every element of python collection like List and transfrom every element as per function.

- But Map function returns map Object , and then we write it as list() to convert the map object into list.

````python
# Map()
map_result=map(sum,l)
list(map_result)
````

````python
# Map() + lambda
map_result2=map(lambda n:(n*(n+1))/2,l)
list(map_result2)
````

### 3) Filter( ) Function

- What problem does it solve ?<br> So If we have Orders Data with order_Status as COMPLETE,PENDING,CLOSED and PROCESSING, We have to create is like :

``` python
    orders = [
    "1,2013-07-25 00:00:00.0,11599,CLOSED",
    "2,2013-07-25 00:00:00.0,256,PENDING",
    "3,2013-07-25 00:00:00.0,12111,COMPLETE",
    "4,2013-07-26 00:00:00.0,8827,COMPLETE"
    ]

    result=[]
    for order in orders:
        if order.split(",")[3]=="COMPLETE":
            result.append(order) 
```

**INSTEAD OF THIS, We can do :**

``` python

    result=filter(lambda order:order.split(",")[3]=="COMPLETE",orders)
    print(list(result))
```






````python
result=filter(lambda order:order.split(",")[3]=="COMPLETE",orders)
print(list(result))
````

````python
# lets create a new list for the order statuses only
result=map(lambda order:order.split(",")[3],orders)
print(list(result))

# now lets print unique order status
set(map(lambda order:order.split(",")[3],orders))
````

### 4) Sorted( ) Function

- Why do we need sorted() function ? <br> So We always need to sort the data, like if we have list num=[1,5,6,4] then we can easily sort it as num=[1,4,5,6] , but what about the string data with mutiple values ?<br>
For this we use a function called Sorted(data,key) where key is the function that tells the column on which sorting is performed


````python
# Sorting done using column 2 timeline
sorted(orders,key=lambda order:order.split(",")[2])
````
